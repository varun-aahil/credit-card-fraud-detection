```python
df.dropna(inplace=True)
```

```python
df['Class'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
X = df.drop('Class',axis=1)
y = df['Class']

X_Scaled = sc.fit_transform(X)
```

```python
from sklearn.ensemble import IsolationForest
```

```python
iso_forest = IsolationForest(n_estimators=200, contamination= 0.005, random_state=42)
```

```python
iso_forest.fit(X_Scaled)
```

```python
y_pred_raw = iso_forest.predict(X_Scaled)
y_pred = np.where(y_pred_raw == -1,1,0)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
print(confusion_matrix(y,y_pred))
```

```python
print(classification_report(y,y_pred))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Convert to pure NumPy arrays to avoid Pandas slicing errors
X_arr = np.array(X)
y_arr = np.array(y)

plt.figure(figsize=(10, 6))

# Plot normal transactions in blue
plt.scatter(X_arr[y_arr == 0, 0], X_arr[y_arr == 0, 1], c='blue', label='Actual Normal', alpha=0.3, s=10)

# Plot predicted anomalies in red
plt.scatter(X_arr[y_pred == 1, 0], X_arr[y_pred == 1, 1], c='red', label='Predicted Fraud', marker='x', s=50)

plt.title('Isolation Forest: Anomaly Detection')
plt.xlabel('Time(scaled)')
plt.ylabel('V1(sclawd)')
plt.legend()
plt.show()
```
